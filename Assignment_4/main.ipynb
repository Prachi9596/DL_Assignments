```python
# ============================================================
# LSTM-BASED TIME-SERIES FORECASTING
# Stock Price Prediction using Apple (AAPL) Dataset
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout


# ============================================================
# 2. DOWNLOAD STOCK PRICE DATASET
# ============================================================

# Download Apple's historical stock data
df = yf.download(
    "AAPL",
    start="2015-01-01",
    end="2025-01-01",
    auto_adjust=False
)

print("Dataset loaded successfully!")
print("\nFirst 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)


# ============================================================
# 3. CHECK DATASET INFORMATION
# ============================================================

print("\nDataset Information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 4. SELECT CLOSE PRICE
# ============================================================

# We only need the Closing Price for forecasting

data = df[["Close"]].values

print("\nClose price data:")
print(data[:5])


# ============================================================
# 5. PLOT ORIGINAL STOCK PRICE
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(data)

plt.title("Apple (AAPL) Historical Closing Price")
plt.xlabel("Days")
plt.ylabel("Closing Price")

plt.show()


# ============================================================
# 6. NORMALIZE THE DATA
# ============================================================

# Convert prices to values between 0 and 1

scaler = MinMaxScaler(feature_range=(0, 1))

scaled_data = scaler.fit_transform(data)

print("\nFirst 5 normalized values:")
print(scaled_data[:5])


# ============================================================
# 7. CREATE TIME-SERIES SEQUENCES
# ============================================================

# The model will use the previous 60 days
# to predict the next day's closing price.

time_step = 60

X = []
y = []

for i in range(time_step, len(scaled_data)):

    # Previous 60 days
    X.append(
        scaled_data[i - time_step:i, 0]
    )

    # Next day's price
    y.append(
        scaled_data[i, 0]
    )


X = np.array(X)
y = np.array(y)

print("\nSequence data:")
print("X shape:", X.shape)
print("y shape:", y.shape)


# ============================================================
# 8. SPLIT DATA INTO TRAINING AND TESTING
# ============================================================

# 80% data for training
# 20% data for testing

train_size = int(len(X) * 0.8)

X_train = X[:train_size]
X_test = X[train_size:]

y_train = y[:train_size]
y_test = y[train_size:]

print("\nTraining data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)


# ============================================================
# 9. RESHAPE DATA FOR LSTM
# ============================================================

# LSTM expects data in the form:
# [samples, time steps, features]

X_train = X_train.reshape(
    X_train.shape[0],
    X_train.shape[1],
    1
)

X_test = X_test.reshape(
    X_test.shape[0],
    X_test.shape[1],
    1
)

print("\nAfter reshaping:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)


# ============================================================
# 10. BUILD LSTM MODEL
# ============================================================

model = Sequential()

# First LSTM layer
model.add(
    LSTM(
        units=50,
        return_sequences=True,
        input_shape=(time_step, 1)
    )
)

# Dropout to reduce overfitting
model.add(
    Dropout(0.2)
)

# Second LSTM layer
model.add(
    LSTM(
        units=50,
        return_sequences=False
    )
)

# Dropout
model.add(
    Dropout(0.2)
)

# Dense layer
model.add(
    Dense(25)
)

# Output layer
model.add(
    Dense(1)
)


# ============================================================
# 11. COMPILE THE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

print("\nModel Summary:")
model.summary()


# ============================================================
# 12. TRAIN THE MODEL
# ============================================================

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)


# ============================================================
# 13. MAKE PREDICTIONS
# ============================================================

predicted_price = model.predict(X_test)


# ============================================================
# 14. CONVERT NORMALIZED VALUES BACK
#    TO ORIGINAL STOCK PRICES
# ============================================================

predicted_price = scaler.inverse_transform(
    predicted_price
)

actual_price = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)


# ============================================================
# 15. CALCULATE RMSE
# ============================================================

rmse = np.sqrt(
    mean_squared_error(
        actual_price,
        predicted_price
    )
)


# ============================================================
# 16. CALCULATE MAE
# ============================================================

mae = mean_absolute_error(
    actual_price,
    predicted_price
)


print("\n======================================")
print("MODEL EVALUATION")
print("======================================")

print("RMSE:", rmse)
print("MAE :", mae)


# ============================================================
# 17. PLOT ACTUAL VS PREDICTED PRICE
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    actual_price,
    label="Actual Price"
)

plt.plot(
    predicted_price,
    label="Predicted Price"
)

plt.title(
    "AAPL Stock Price - Actual vs Predicted"
)

plt.xlabel("Time")
plt.ylabel("Stock Price")

plt.legend()

plt.show()


# ============================================================
# 18. PLOT TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()

plt.show()


# ============================================================
# 19. PREDICT NEXT DAY'S STOCK PRICE
# ============================================================

# Take the last 60 days of available data

last_60_days = scaled_data[-time_step:]


# Reshape the data for LSTM

last_60_days = last_60_days.reshape(
    1,
    time_step,
    1
)


# Predict next day

next_day_prediction = model.predict(
    last_60_days
)


# Convert prediction back to original price

next_day_prediction = scaler.inverse_transform(
    next_day_prediction
)


print("\n======================================")
print("NEXT DAY STOCK PRICE FORECAST")
print("======================================")

print(
    "Predicted next day's closing price:",
    float(next_day_prediction[0][0])
)
```
